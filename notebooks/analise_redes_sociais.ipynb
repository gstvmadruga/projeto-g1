## 1. Introdução ao problema

Este projeto tem como objetivo analisar dados de redes sociais brasileiras, comparando plataformas, perfis, categorias e tipos de conteúdo por meio de métricas como seguidores, curtidas, comentários, compartilhamentos, visualizações, alcance e taxa de engajamento.

A análise busca identificar padrões de desempenho entre os diferentes elementos presentes na base de dados, utilizando técnicas de análise exploratória e visualização de dados com Python.

Os resultados serão utilizados para desenvolver indicadores (KPIs), gráficos e um dashboard interativo que facilite a interpretação das informações.

## 2. Explicação da base de dados

A base de dados utilizada neste projeto contém informações simuladas sobre publicações realizadas em diferentes redes sociais no Brasil.

Cada registro representa uma publicação e apresenta informações relacionadas à plataforma utilizada, perfil, categoria, tipo de conteúdo, quantidade de seguidores e métricas de desempenho.

Entre as principais informações disponíveis estão:

- Ano, mês e data da publicação;
- Plataforma da rede social;
- Perfil responsável pela publicação;
- Categoria do conteúdo;
- Tipo de conteúdo;
- Número de seguidores;
- Curtidas;
- Comentários;
- Compartilhamentos;
- Visualizações;
- Alcance;
- Taxa de engajamento;
- Horário da publicação.

Essas informações serão utilizadas para realizar a análise exploratória, calcular indicadores (KPIs) e criar visualizações que permitam comparar o desempenho das diferentes redes sociais e conteúdos.

## 3. Leitura dos dados

Nesta etapa, a base de dados será carregada utilizando a biblioteca Pandas, permitindo que os dados sejam analisados e posteriormente utilizados nas visualizações e no dashboard.

```python
import pandas as pd

df = pd.read_csv("../dados/simulacao_redes_sociais_brasil (1).csv")

df.head()
```

## 4. Limpeza e preparação dos dados

Nesta etapa, os dados serão preparados para a análise. A coluna de data será convertida para o formato de data do Pandas, facilitando análises temporais e a criação de visualizações ao longo do projeto.

```python
df["data"] = pd.to_datetime(df["data"])

df.head()
```

```python
from sqlalchemy import create_engine
import os

os.makedirs("../database", exist_ok=True)

engine = create_engine("sqlite:///../database/redes_sociais.db")

df.to_sql(
    "publicacoes",
    engine,
    if_exists="replace",
    index=False
)

print("Dados salvos no banco de dados com sucesso!")
```

## 5. Engenharia de atributos

Para melhorar a análise dos dados, será criada uma nova métrica chamada `interacoes`.

Essa métrica representa a soma das curtidas, comentários e compartilhamentos de cada publicação, permitindo analisar de forma conjunta as principais interações realizadas pelos usuários.

```python
df["interacoes"] = (
    df["curtidas"]
    + df["comentarios"]
    + df["compartilhamentos"]
)

df[["curtidas", "comentarios", "compartilhamentos", "interacoes"]].head()
```

## 6. Análise exploratória

Nesta etapa, serão analisados os dados para identificar padrões de desempenho entre as plataformas, tipos de conteúdo e perfis presentes na base.

```python
df.groupby("plataforma")["taxa_engajamento"].mean()
```

### Análise por tipo de conteúdo

Será analisada a taxa média de engajamento de cada tipo de conteúdo presente na base, permitindo comparar o desempenho dos diferentes formatos de publicação.

```python
df.groupby("tipo_conteudo")["taxa_engajamento"].mean()
```

### Análise por perfil

Será analisada a taxa média de engajamento de cada perfil presente na base de dados, permitindo identificar quais perfis apresentam melhor desempenho nas publicações.

```python
df.groupby("perfil")["taxa_engajamento"].mean()
```

### Conclusão da análise exploratória

A análise exploratória permite observar diferenças na taxa média de engajamento entre plataformas, tipos de conteúdo e perfis. Essas comparações ajudam a identificar quais elementos apresentam melhor desempenho e servirão como base para a definição dos principais indicadores (KPIs) do projeto.

## 7. KPIs

Nesta etapa, serão calculados os principais indicadores de desempenho (KPIs) da base de dados. Esses indicadores serão utilizados para resumir os resultados e facilitar a interpretação do desempenho das publicações nas redes sociais.

```python
kpis = {
    "Total de publicações": len(df),
    "Total de seguidores": df["seguidores"].sum(),
    "Total de curtidas": df["curtidas"].sum(),
    "Total de comentários": df["comentarios"].sum(),
    "Total de compartilhamentos": df["compartilhamentos"].sum(),
    "Total de visualizações": df["visualizacoes"].sum(),
    "Taxa média de engajamento": df["taxa_engajamento"].mean()
}

kpis
```

## 8. Gráficos

Nesta etapa, serão criadas visualizações para representar os resultados encontrados na análise dos dados. Os gráficos permitirão comparar o desempenho das plataformas e dos diferentes tipos de conteúdo.

```python
import matplotlib.pyplot as plt
import seaborn as sns

engajamento_plataforma = df.groupby("plataforma")["taxa_engajamento"].mean().sort_values(ascending=False)

plt.figure(figsize=(10, 5))
sns.barplot(x=engajamento_plataforma.index, y=engajamento_plataforma.values)

plt.title("Taxa Média de Engajamento por Plataforma")
plt.xlabel("Plataforma")
plt.ylabel("Taxa Média de Engajamento (%)")
plt.xticks(rotation=45)
plt.show()
```

```python
engajamento_ano = df.groupby("ano")["taxa_engajamento"].mean()

plt.figure(figsize=(10, 5))
plt.plot(engajamento_ano.index, engajamento_ano.values, marker="o")

plt.title("Evolução da Taxa Média de Engajamento")
plt.xlabel("Ano")
plt.ylabel("Taxa Média de Engajamento (%)")
plt.grid(True)
plt.show()
```

```python
plt.figure(figsize=(10, 5))

plt.scatter(
    df["seguidores"],
    df["taxa_engajamento"],
    alpha=0.5
)

plt.title("Relação entre Seguidores e Taxa de Engajamento")
plt.xlabel("Número de Seguidores")
plt.ylabel("Taxa de Engajamento (%)")
plt.grid(True)
plt.show()
```

### Análise de correlação

A análise de correlação será utilizada para verificar a relação entre as principais métricas numéricas presentes na base de dados. Essa análise permite identificar quais variáveis apresentam maior ou menor relação entre si.

```python
correlacao = df[
    [
        "seguidores",
        "curtidas",
        "comentarios",
        "compartilhamentos",
        "visualizacoes",
        "alcance",
        "taxa_engajamento"
    ]
].corr()

correlacao
```

### Mapa de correlação

O mapa de correlação apresenta visualmente a relação entre as principais métricas numéricas da base de dados. Valores próximos de 1 indicam uma relação positiva forte, enquanto valores próximos de -1 indicam uma relação negativa forte.

```python
plt.figure(figsize=(10, 6))

sns.heatmap(
    correlacao,
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.title("Mapa de Correlação das Métricas")
plt.show()
```
